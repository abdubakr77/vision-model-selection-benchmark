# Documents: data preparation

**Phase:** Phase 3

**Purpose:** Build ONE canonical dataset and a fixed split that every model reads.

**Inputs:** Raw dataset, EDA decisions.

**Outputs:** split.csv (image_id, split, label, path + dataset-specific columns), meta.json, preview image, optional zip for reuse as a Kaggle dataset.

## Steps

- [ ] Download or locate raw data
- [ ] Resize to the canonical size, save images
- [ ] Create train / val / test split with seed 42
- [ ] Write split.csv and meta.json
- [ ] Visual check of labels on a preview grid
- [ ] Define the shared task (field extraction or QA) and its metric
