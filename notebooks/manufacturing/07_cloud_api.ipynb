# Cloud vision model (API)

**Phase:** Phase 2

**Purpose:** Cloud-hosted VLM called through an API on the test split only.

**Inputs:** `bench/split.csv` and images of VisA pcb1 from `02_data_prep`.

**Outputs:** results/per_image/<dataset>_<model>.csv (image_id, prediction, score, latency_ms, vram_mb, raw_output, parse_ok, tokens_in, tokens_out) and a summary row.

## Steps

- [ ] Load split.csv and the model (try/except fallback)
- [ ] Train on train split (supervised models only)
- [ ] Tune threshold or prompt on val split only
- [ ] Run on test split, log every image
- [ ] Record latency (after warm-up), peak VRAM, model size
- [ ] Free GPU memory before the next model
- [ ] API key in Kaggle Secrets, never in the notebook
- [ ] Cache every response
- [ ] Log token usage and compute cost per 1,000 images from public prices
